## ✅ PROJECT 1 — Data Collection & Cleaning
> **Goal:** Load a real-world dataset, understand its structure, remove duplicates, handle missing values, and fix formatting issues.

### 📦 Cell 1 — Import Libraries & Load Dataset

```python
# ─────────────────────────────────────────────
#  CELL 1 | Import Libraries & Load Dataset
# ─────────────────────────────────────────────

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Load Titanic dataset directly from URL (no file download needed)
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

# Standardize column names to lowercase right away
df.columns = df.columns.str.lower().str.strip()

print("✅ Dataset loaded successfully!")
print(f"   Shape: {df.shape[0]} rows × {df.shape[1]} columns")
print(f"   Columns: {df.columns.tolist()}")
df.head()
```

### 🔍 Cell 2 — Inspect the Dataset

```python
# ─────────────────────────────────────────────
#  CELL 2 | Inspect the Dataset
# ─────────────────────────────────────────────

print("━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")
print(" Dataset Info (columns, types, non-null)")
print("━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")
df.info()

print("\n━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")
print(" Missing Values Per Column")
print("━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")
missing = df.isnull().sum()
missing_pct = (df.isnull().sum() / len(df) * 100).round(2)
missing_df = pd.DataFrame({'Missing Count': missing, 'Missing %': missing_pct})
print(missing_df[missing_df['Missing Count'] > 0])

print("\n━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")
print(f" Duplicate Rows: {df.duplicated().sum()}")
print("━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")
```

### 🧹 Cell 3 — Clean the Dataset

```python
# ─────────────────────────────────────────────
#  CELL 3 | Clean the Dataset
# ─────────────────────────────────────────────

print(f"Shape before cleaning: {df.shape}")

# Step 1 — Remove duplicate rows
df = df.drop_duplicates()
print(f"✔ Duplicates removed")

# Step 2 — Handle missing values
df['age'].fillna(df['age'].median(), inplace=True)          # Numeric → fill with median
df['embarked'].fillna(df['embarked'].mode()[0], inplace=True) # Categorical → fill with mode
df.drop(columns=['cabin'], inplace=True)                      # >77% missing → drop column
print("✔ Missing values handled")

# Step 3 — Fix formatting
df['name'] = df['name'].str.strip()   # Remove leading/trailing spaces
df['sex'] = df['sex'].str.lower()     # Ensure lowercase
print("✔ Formatting fixed")

# Confirm clean dataset
print(f"\nShape after cleaning: {df.shape}")
print(f"Remaining nulls: {df.isnull().sum().sum()}")

# Save cleaned dataset
df.to_csv('cleaned_titanic.csv', index=False)
print("\n✅ Cleaned dataset saved as: cleaned_titanic.csv")
df.head()
```

## ✅ PROJECT 2 — Exploratory Data Analysis (EDA)
> **Goal:** Calculate statistics, find patterns, detect outliers, understand correlations, and create visualizations.

### 📐 Cell 4 — Basic Statistics

```python
# ─────────────────────────────────────────────
#  CELL 4 | Basic Statistics
# ─────────────────────────────────────────────

print("━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")
print(" Descriptive Statistics")
print("━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")
print(df[['age', 'fare', 'pclass']].describe().round(2))

print("\n━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")
print(" Survival Rate")
print("━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")
print(f"  Overall:    {df['survived'].mean()*100:.1f}% survived")
print(f"  Female:     {df[df['sex']=='female']['survived'].mean()*100:.1f}% survived")
print(f"  Male:       {df[df['sex']=='male']['survived'].mean()*100:.1f}% survived")
print(f"  1st Class:  {df[df['pclass']==1]['survived'].mean()*100:.1f}% survived")
print(f"  3rd Class:  {df[df['pclass']==3]['survived'].mean()*100:.1f}% survived")
```

### 🔎 Cell 5 — Outliers & Correlations

```python
# ─────────────────────────────────────────────
#  CELL 5 | Outliers & Correlations
# ─────────────────────────────────────────────

# --- Detect Outliers using IQR method ---
Q1  = df['fare'].quantile(0.25)
Q3  = df['fare'].quantile(0.75)
IQR = Q3 - Q1

outliers = df[(df['fare'] < Q1 - 1.5 * IQR) | (df['fare'] > Q3 + 1.5 * IQR)]

print("━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")
print(" Outlier Detection — Fare Column (IQR)")
print("━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")
print(f"  Q1  = {Q1:.2f}  |  Q3  = {Q3:.2f}  |  IQR = {IQR:.2f}")
print(f"  Lower fence = {Q1 - 1.5*IQR:.2f}")
print(f"  Upper fence = {Q3 + 1.5*IQR:.2f}")
print(f"  Outliers found: {len(outliers)} rows")

# --- Correlation Matrix ---
print("\n━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")
print(" Correlation Matrix")
print("━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")
corr = df[['survived', 'age', 'fare', 'pclass']].corr().round(3)
print(corr)
```

### 📊 Cell 6 — Visualizations

```python
# ─────────────────────────────────────────────
#  CELL 6 | Visualizations
# ─────────────────────────────────────────────

sns.set_theme(style="whitegrid", palette="Set2")
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle("Titanic Dataset — EDA Visualizations", fontsize=16, fontweight='bold', y=1.01)

# --- Plot 1: Survival Count (Bar Chart) ---
sns.countplot(
    data=df, x='survived', ax=axes[0, 0],
    palette=['#e74c3c', '#2ecc71']
)
axes[0, 0].set_title('Survival Count', fontweight='bold')
axes[0, 0].set_xlabel('Survived (0 = No, 1 = Yes)')
axes[0, 0].set_ylabel('Count')
axes[0, 0].bar_label(axes[0, 0].containers[0])

# --- Plot 2: Age Distribution (Histogram) ---
axes[0, 1].hist(df['age'], bins=30, color='steelblue', edgecolor='white', linewidth=0.6)
axes[0, 1].set_title('Age Distribution', fontweight='bold')
axes[0, 1].set_xlabel('Age')
axes[0, 1].set_ylabel('Count')
axes[0, 1].axvline(df['age'].median(), color='red', linestyle='--', label=f"Median: {df['age'].median():.0f}")
axes[0, 1].legend()

# --- Plot 3: Avg Fare by Passenger Class (Line Graph) ---
avg_fare = df.groupby('pclass')['fare'].mean()
axes[1, 0].plot(avg_fare.index, avg_fare.values, marker='o', color='coral', linewidth=2, markersize=8)
axes[1, 0].set_title('Avg Fare by Passenger Class', fontweight='bold')
axes[1, 0].set_xlabel('Passenger Class')
axes[1, 0].set_ylabel('Average Fare ($)')
axes[1, 0].set_xticks([1, 2, 3])
for x, y in zip(avg_fare.index, avg_fare.values):
    axes[1, 0].annotate(f'${y:.1f}', (x, y), textcoords='offset points', xytext=(0, 10), ha='center')

# --- Plot 4: Correlation Heatmap ---
corr_matrix = df[['survived', 'age', 'fare', 'pclass']].corr()
sns.heatmap(
    corr_matrix, annot=True, fmt='.2f',
    cmap='coolwarm', ax=axes[1, 1],
    linewidths=0.5, square=True
)
axes[1, 1].set_title('Correlation Heatmap', fontweight='bold')

plt.tight_layout()
plt.savefig('eda_plots.png', dpi=150, bbox_inches='tight')
plt.show()
print("✅ Saved: eda_plots.png")
```

### 🎯 Cell 7 — Key Insights (Summary)

```python
# ─────────────────────────────────────────────
#  CELL 7 | Key Insights Summary
# ─────────────────────────────────────────────

print("━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")
print("  KEY INSIGHTS — Titanic EDA")
print("━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")

survival_rate  = df['survived'].mean() * 100
female_surv    = df[df['sex']=='female']['survived'].mean() * 100
male_surv      = df[df['sex']=='male']['survived'].mean() * 100
p1_surv        = df[df['pclass']==1]['survived'].mean() * 100
p3_surv        = df[df['pclass']==3]['survived'].mean() * 100
median_age     = df['age'].median()
avg_fare_all   = df['fare'].mean()

print(f"  1. Overall survival rate    : {survival_rate:.1f}%")
print(f"  2. Female survival rate     : {female_surv:.1f}% (much higher than male)")
print(f"  3. Male survival rate       : {male_surv:.1f}%")
print(f"  4. 1st Class survival       : {p1_surv:.1f}% vs 3rd Class: {p3_surv:.1f}%")
print(f"  5. Median passenger age     : {median_age:.0f} years")
print(f"  6. Average fare paid        : ${avg_fare_all:.2f}")
print(f"  7. Fare outliers detected   : {len(outliers)} high-paying passengers")
print("━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")
print("  ✅ Both Project 1 & 2 complete. Ready to submit!")
print("━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")
```
